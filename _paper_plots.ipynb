```python
import os
import numpy             as np
import matplotlib.pyplot as plt
import pandas            as pd 
from matplotlib.ticker import (AutoMinorLocator, MultipleLocator)


folder = './plots/paper/'
if not os.path.exists(folder):
    os.makedirs(folder)
```

```python
our_color = '#C47B7B'
sdjwt_color = '#87A987'
pjv_color= '#779ECB'
```

```python
our_marker = '+'
pjv_marker = 's'
sdjwt_marker = 'x'
```

```python
marker_range_10 = range(0,10,1)
marker_range_100 = range(0,100,10)
```

# Iterating over permissions

### Iterating over permissions - DC Issuance Latency

```python
# Latencies in issuing DCs when iterating over permissions. Each credential was delegated 10 times. 
# This graph plots the length of the 10th delegated credential when the number of permission varies between 1 and 10.
df = pd.read_csv("./csv_dir/iop_vc_issuance.csv")
fig, ax = plt.subplots()
x = range(1,11,1)

ax.plot(x, df['Ours'] / 1000, label='Ours', color=our_color, marker=our_marker, markevery=marker_range_10, linewidth=2)
ax.plot(x, df['PJVs'] / 1000, label='Vrielynck et al.', color=pjv_color, marker=pjv_marker, markevery=marker_range_10, linewidth=2)
ax.plot(x, df['SdJWT'] / 1000, label='Delegate SD-JWT', color=sdjwt_color, marker=sdjwt_marker, markevery=marker_range_10, linewidth=2)

ax.xaxis.set_major_locator(MultipleLocator(1))
ax.xaxis.set_minor_locator(AutoMinorLocator(5))
ax.yaxis.set_major_locator(MultipleLocator(0.1))
ax.yaxis.set_minor_locator(AutoMinorLocator(5))

ax.set_xlabel("Number of permissions")
ax.set_ylabel("Latency (ms)")

ax.legend(loc="upper left", fancybox=True, framealpha=0.4, prop={'size': 9})

plt.xlim(0.9, 10.1)
plt.ylim(0, 1)

plt.grid(True, which='both', linestyle=':', linewidth=0.5)
plt.savefig(f"{folder}/Iterating_Over_Permissions_VC_Issuance.pdf", dpi=600, format='pdf')
plt.show()
```
